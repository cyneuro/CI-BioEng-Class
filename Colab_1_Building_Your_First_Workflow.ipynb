### Topic: The General Pattern for Accessing National Research Computing

Some research questions need more computing power than a laptop or Google Colab can provide. Getting access to that kind of power almost always follows the same general pattern, no matter which specific system you end up using: **create an account, get approved for a project, generate a security credential, and confirm you're connected.** FABRIC — a national research testbed — is this course's worked example, but the pattern transfers directly to your own institution's HPC cluster or any other system your research group uses. ### 🧭 Activity A.1 — Understanding Your Compute Environments

Before requesting any real computing resources, it's worth knowing what you're actually choosing between. Three environments come up throughout this course:

| Environment | What it actually is | Reaches FABRIC directly? | Best for |
|---|---|---|---|
| **Google Colab** | A free, hosted Jupyter notebook in your browser. Nothing to install; nothing persists on its own. | ❌ No dependable network path into a research testbed | Quick, shareable, zero-install work |
| **A testbed's own JupyterHub** (e.g., FABRIC's) | A multi-user Jupyter deployment the testbed itself runs, inside its own network, with its toolkit and your credentials already configured. | ✅ Yes — that's the whole point of running it there | Provisioning, running, and releasing real computing sessions |
| **VSCode (with Remote-SSH)** | A full code editor that can open a direct terminal/file connection to a remote machine, including a running compute node. | ✅ Yes, once you already have a running session to connect to | Serious day-to-day development once a session already exists |
