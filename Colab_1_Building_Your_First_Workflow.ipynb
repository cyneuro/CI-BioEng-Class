### 🧭 Activity A.0 — The CI-Hardware Tier Ladder: From Laptop to National Resource

Before comparing *how* you access compute (Activity A.1, next), it's worth knowing the actual scale differences between tiers of hardware — with real numbers, not just "bigger."

| Tier | Typical scale | Who it serves |
|---|---|---|
| **micro:bit** | 128 KB of RAM total | One device, one task (you'll work with this directly in Colab 3) |
| **Laptop** | ~8–16 CPU cores, 16–32 GB RAM | One person |
| **Local/departmental server** | ~32–128 cores, 256 GB–1 TB RAM, often 1–8 GPUs | One lab or department, shared |
| **Institutional HPC cluster ("local mainframe")** | Hundreds to thousands of cores across many networked machines, dozens to hundreds of GPUs | An entire university, shared and scheduled |
| **National resource (FABRIC)** | A testbed distributed across dozens of sites nationwide, built for large-scale, reconfigurable experimentation | Researchers nationwide, by project allocation |
| **The broader ecosystem (ACCESS, formerly XSEDE)** | Coordinates allocations across *multiple* national supercomputing centers at once, collectively offering computing power measured in petaflops (quadrillions of operations per second) | All NSF-supported research nationally |
| **NAIRR (National AI Research Resource)** | A newer, AI-specific national resource, launched specifically to give researchers access to the GPU-heavy infrastructure large AI models require | AI/ML researchers nationally |

That's the difference between a 5-minute laptop job and a job that either ties up your laptop for over an hour, or finishes in a few minutes on FABRIC's GPU-enabled nodes instead. ### 🧭 Activity A.2 — The CI-Software Abstraction Ladder: From Machine Code to Vibe Coding

Every AI Prompt Challenge in this course sits at the current endpoint of a real, decades-long historical progression: `machine code (binary) → assembly → compiled high-level languages (Fortran, C, 1950s–70s) → interpreted/scripting languages with automatic memory management (Python) → frameworks and libraries → vibe coding, today`. Each step handed a layer of low-level detail over to a machine, freeing the human to work at a higher level of intent.

**A useful piece of history:** when compilers first appeared in the 1950s, experienced assembly programmers were skeptical of them for the same reasons some developers are skeptical of AI-assisted coding today — trusting a layer of abstraction you didn't write yourself. This isn't a new anxiety unique to AI; it's a recurring pattern every time computing adds a layer of abstraction.

### Concept 1 — Software Drift



**Resource:** A workflow that runs correctly today can fail months later even with unchanged code and data, because one or more of the libraries it depends on quietly updated in the meantime. This is <em>software drift</em> — the same script, a different result, because the ground it's standing on shifted.

```python
def old_behavior(x):

    return x // 2          # integer (floor) division -- an older library's default



def new_behavior(x):

    return x / 2            # true division -- the same call, a newer library's default



print("Old library:", old_behavior(5))   # 2    (drifted!)

print("New library:", new_behavior(5))   # 2.5
```

### Concept 2 — Stage Independence



**Resource:** A well-designed workflow keeps its stages — obtaining data, analyzing it, visualizing it, documenting it — independent of one another. <code>run_anova()</code> in this Colab doesn't know or care whether <code>get_data()</code> fetched its input from a URL, a local file, or a simulated array.

```python
import pandas as pd

from scipy import stats



def run_anova_demo(data):

    groups = [g["value"].values for _, g in data.groupby("condition")]

    f_stat, p_val = stats.f_oneway(*groups)

    return f_stat, p_val



from_a_url = pd.DataFrame({"condition": ["a","a","b","b"], "value": [1,2,5,6]})

from_a_simulation = pd.DataFrame({"condition": ["a","a","b","b"], "value": [10,11,2,3]})

print(run_anova_demo(from_a_url))

print(run_anova_demo(from_a_simulation))
```

### Concept 4 — Provenance/Documentation Logging



**Resource:** A workflow that saves a small log alongside its results — what date it ran, what settings were used, how many results came out significant — lets you (or anyone else) reconstruct exactly what happened later. Without this, reproducing a result means guessing.

```python
from datetime import date

import pandas as pd



def document_research(anova_result):

    n_sig = int(anova_result["significant"].sum())

    return pd.DataFrame([{"date": date.today().isoformat(), "n_significant": n_sig}])
```

### Concept 5 — Deterministic Workflows



**Resource:** A deterministic workflow, given the exact same input and settings, always produces the exact same output. This is what lets a collaborator trust that re-running your analysis will get them the same conclusions you got.

```python
import numpy as np

rng1 = np.random.default_rng(seed=42)

rng2 = np.random.default_rng(seed=42)

print(rng1.normal(size=3))

print(rng2.normal(size=3))  # identical -- the fixed seed makes this deterministic
```
