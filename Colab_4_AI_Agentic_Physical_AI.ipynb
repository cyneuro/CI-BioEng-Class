### Concept 2 — Machine Learning



**Resource:** Machine learning is a subset of AI where a system improves at a task by learning patterns from data, rather than following only explicitly hand-written rules — the trained-classifier half of Activity A.1's comparison.

```python
from sklearn.linear_model import LogisticRegression

import numpy as np

rng = np.random.default_rng(0)

X = rng.normal(0, 1, size=(50, 2))

y = (X[:, 0] + X[:, 1] > 0).astype(int)

clf = LogisticRegression().fit(X, y)   # "learning" the pattern from data

print("Learned coefficients:", clf.coef_)
```

### Concept 7 — Perception-Decision-Action Loop



**Resource:** The standard cycle in physical or agentic AI: sense the environment, decide what to do, act, then sense the result and repeat — this Colab's control network (Part B) and Cutebot execution (Part C) together form exactly this loop.

```python
def sense(v): return v

def decide(v): return "increase" if v < 50 else "decrease"

def act(v, a): return v + 5 if a == "increase" else v - 5



v = 30

for _ in range(3):

    perceived = sense(v)

    action = decide(perceived)

    v = act(v, action)

    print(f"perceived={perceived}, action={action}, new_value={v}")
```
