The functions below read data from a single subject and combine the resulting dataset into an array of size (subjects × game_types × channels × time). These are provided for you — no AI prompt needed.

```python
import numpy as np



import pandas as pd







def read_subject(subject, path):



    assert len(os.listdir(path)) == 4







    subject_recordings = []



    for game_idx in range(4):



        # Read channel x time data



        csv_filename = f'{subject}G{game_idx + 1}AllChannels.csv'



        df = pd.read_csv(os.path.join(path, csv_filename))



        df.drop("Unnamed: 14", axis = 1, inplace = True)







        # Shape into a channels x time array



        subject_recordings.append(df.to_numpy().T)







    # Shape into game_types x channels x time



    subject_recordings = np.stack(subject_recordings)







    return subject_recordings







def read_subjects(subjects, path):



    all_recordings = []



    for subject in subjects:



        subject_path = os.path.join(path, f"({subject})", "Preprocessed EEG Data", ".csv format")



        data = read_subject(subject, subject_path)



        all_recordings.append(data)







    # Subjects x game_types x channels x time



    all_recordings = np.stack(all_recordings)



    return all_recordings
```

```python
data = read_subjects(subjects, path)
```

### Concept 5 — Checksum Verification



**Resource:** A checksum is a short fingerprint computed from a file's contents — if you download the same dataset twice and get different checksums, the file changed or got corrupted in transit, something a file size alone wouldn't necessarily catch.

```python
import hashlib

data_v1 = b"subject data version 1"

data_v2 = b"subject data version 1 (corrupted)"

print("Checksum 1:", hashlib.md5(data_v1).hexdigest())

print("Checksum 2:", hashlib.md5(data_v2).hexdigest())
```

### Concept 6 — Speedup Factor



**Resource:** If splitting work evenly across N processors cuts total runtime by a factor of N, that's called linear (ideal) speedup — real speedup is usually less than N because of coordination overhead.

```python
serial_time = 100  # seconds

parallel_time = 25  # seconds using 4 cores

speedup = serial_time / parallel_time

print(f"Speedup factor: {speedup}x")
```

### Concept 8 — Per-Participant Error Isolation



**Resource:** When one subject's data is malformed in a 28-subject batch run, a well-designed workflow logs that one failure and continues processing the other 27, rather than the entire run crashing.

```python
def process_subject(subject_id, subjects_with_errors):

    if subject_id in subjects_with_errors:

        print(f"[LOGGED ERROR] {subject_id} failed -- continuing with others.")

        return None

    return f"{subject_id}: processed successfully"



for s in ["S01", "S02", "S26"]:

    print(process_subject(s, subjects_with_errors={"S26"}))
```
